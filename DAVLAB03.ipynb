```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully")
```

```python
data={"Student":["Asha","Bharat","Chitra","Deepak","Esha","Farhan",
                 "Gowri","Harish","Isha","Jatin","Kavya","Lokesh",
                 "Meera","Nikhil","Pooja","Rahul","Sneha","Tarun","Uma","Varun"],
      "Department":["CSE","AIML","CSE","ISE","AIML","CSE","ISE"
      ,"CSE","AIML","ISE","CSE","AIML",
                    "ISE","CSE","AIML","ISE","CSE","AIML","ISE","CSE"],
      "Attendance":[92,78,85,96,88,72,91,65,87,82,94,76,89,97,84,69,93,80,86,90],
      "Study_Hours":[3.5,2,3,5,4,1.5,4.5,1,3.5,2.5,5.5,2,4,6,3,1.5,5,2.5,3.5,4.5],
      "Assignment_Score":[82,68,78,94,86,62,90,55,84,73,96,70,88,98,79,60,93,74,85,91],
      "Internal_Marks":[84,70,80,95,87,65,91,58,85,75,97,72,89,99,81,63,94,76,86,92],
      "Final_Score":[86,72,82,96,88,66,93,60,86,77,98,73,90,99,83,64,95,78,87,94]}
df=pd.DataFrame(data)
display(df.head())
print("Shape:",df.shape)
print(df.dtypes)
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Final_Score"], bins=6,edgecolor="black")
plt.xlabel("Final Score")
plt.ylabel("Frequency")
plt.title("Distribution of Final Scores")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["Final_Score"])
plt.ylabel("Final Score")
plt.title("Boxplot of Final Scores")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x=df["Attendance"],y=df["Final_Score"])
plt.xlabel("Attendance")
plt.ylabel("Final Score")
plt.title("Attendance vs Final Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Study_Hours",y="Final_Score",hue="Department",s=80)
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Study Hours vs Final Score")
plt.show()
```

```python
num_cols=["Attendance","Study_Hours","Assignment_Score","Internal_Marks","Final_Score"]
sns.pairplot(df[num_cols])
plt.show()
```

```python
corr=df[num_cols].corr()
display(corr)
plt.figure(figsize=(8,5))
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()
#
```

```python
movie_data={"Movie":["Sky Force","Code Red","Midnight Run","Ocean Echo","City Lights","Last Game","Silent Road","Future World","Hidden Truth","Wild Journey","Digital Dreams","Final Mission"],"Budget_Crore":[80,120,60,45,35,70,50,150,40,90,110,130],"Duration_Minutes":[145,160,125,135,110,150,130,170,120,155,165,158],"Rating":[8.2,7.5,8,7.8,6.9,8.5,7.6,7.2,8.1,7.9,7.4,8.3],"Revenue_Crore":[240,280,150,110,75,260,130,310,140,220,290,350]}
movies=pd.DataFrame(movie_data)
movies
```

```python
plt.figure(figsize=(8,5))
plt.hist(movies["Rating"], bins=5, edgecolor="black")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.title("Distribution of Movie Ratings")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=movies["Budget_Crore"])
plt.ylabel("Budget (Crore)")
plt.title("Boxplot of Movie Budgets")
plt.show()
```

```python
numerical_cols_movies = ["Budget_Crore", "Duration_Minutes", "Rating", "Revenue_Crore"]
sns.pairplot(movies[numerical_cols_movies])
plt.suptitle("Pair Plot of Numerical Columns in Movie Data", y=1.02)
plt.show()
```

```python
corr_movies = movies[numerical_cols_movies].corr()
display(corr_movies)
plt.figure(figsize=(8,5))
sns.heatmap(corr_movies, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Numerical Columns in Movie Data")
plt.show()
```
